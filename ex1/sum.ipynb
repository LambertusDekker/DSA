#Check Even or Odd

https://www.geeksforgeeks.org/dsa/check-whether-given-number-even-odd/

Given a number n, check whether it is even or odd. Return true for even and false for odd.

```
Examples: 

Input: n = 15
Output: false
Explanation: 15 % 2 = 1, so 15 is odd .

Input: n = 44
Output: true
Explanation: 44 % 2 = 0, so 44 is even.
```


````python
def main():
    n = int(input("give me a number:"))
    r = remainder(n)
    print(f"Remainder: {r}.")
    if r == 0:
        print("even.")
    else:
        print("odd.")

def remainder(n):
    r = n % 2
    return(r)

if __name__ == "__main__":
    main()
````
